```python
from pyspark.sql import functions as F
```

```python
dbutils.widgets.text("catalog", "")
catalog = dbutils.widgets.get("catalog")

spark.sql(f"USE CATALOG {catalog}")
```

```python


# Caminho dos dados
file_path = "/Volumes/analytics_dev/default/analytics/transactions_sales/"

# Leitura
df = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(file_path)
)

# 🔥 Corrigir nomes das colunas (IMPORTANTE)
df = df.toDF(
    "data_venda",
    "produto",
    "categoria",
    "preco_unitario",
    "custo_unitario",
    "marca",
    "qtd_vendida",
    "nome_cliente",
    "localidade"
)

df = df.withColumn("ingestion_time", F.current_timestamp())

# Escrita bronze (dinâmica)
df.write.mode("overwrite").saveAsTable("bronze.transactions_sales")

print("Dados gravados com sucesso")
```
