```python
from pyspark.sql import functions as F
```

```python
# =========================================
# Bronze Layer
# =========================================

# Widgets recebidos do Databricks Bundle
dbutils.widgets.text("catalog", "")
catalog = dbutils.widgets.get("catalog")

# Define catálogo dinamicamente
spark.sql(f"USE CATALOG {catalog}")

# Caminho dos dados
file_path = "/Volumes/analytics_dev/default/analytics/transactions_sales/"

# Leitura CSV
df = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(file_path)
)

# Padronização das colunas
df = df.toDF(
    "data_venda",
    "produto",
    "categoria",
    "preco_unitario",
    "custo_unitario",
    "marca",
    "qtd_vendida",
    "nome_cliente",
    "localidade"
)

# Coluna de ingestão
df = df.withColumn("ingestion_time", F.current_timestamp())

# Escrita Bronze
df.write.mode("overwrite").saveAsTable(
    "bronze.transactions_sales"
)

print("Bronze carregada com sucesso")
```
