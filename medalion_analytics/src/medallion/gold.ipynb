```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
```

```python
##dbutils.widgets.text("catalog", "")
catalog = dbutils.widgets.get("catalog")
```

```python

gold_df = spark.read.table(f"{catalog}.silver.clean_sales")

gold_df = (
    gold_df
    .withColumn("valor_total", F.col("preco_unitario") * F.col("qtd_vendida"))
    .withColumn("custo_total", F.col("custo_unitario") * F.col("qtd_vendida"))
    .withColumn("lucro", F.col("valor_total") - F.col("custo_total"))
)

gold_df.write.mode("overwrite").saveAsTable(
    f"{catalog}.gold.fact_sales"
)
```
